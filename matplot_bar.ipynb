1. Write a Python programming to display a bar chart of the popularity of programming

5. Write a Python programming to display a bar chart of the popularity of programming
Languages. Attach a text label above each bar displaying its popularity (float value).
Sample data:
Programming languages: Java, Python, PHP, JavaScript, C#, C++
Popularity: 22.2, 17.6, 8.8, 8, 7.7, 6.7

```python

languages = ['Java', 'Python', 'PHP', 'JavaScript', 'C#', 'C++']
popularity = [22.2, 17.6, 8.8, 8, 7.7, 6.7]

plt.bar(languages, popularity, color='skyblue')

for i in range(len(languages)):
    plt.text(i, popularity[i] + 0.5, str(popularity[i]), ha='center')  # Label on top

plt.xlabel("Programming Languages")
plt.ylabel("Popularity (%)")
plt.title("Popularity of Programming Languages")
```

10. Write a Python program to create bar plot of scores by group and gender. Use multiple X
values on the same chart for men and women.
Sample Data:
Means (men) = (22, 30, 35, 35, 26)
Means (women) = (25, 32, 30, 35, 29)

```python
men_means = [22, 30, 35, 35, 26]
women_means = [25, 32, 30, 35, 29]
groups = ['A', 'B', 'C', 'D', 'E']

x = np.arange(len(groups))

plt.bar(x - 0.2, men_means, width=0.4, label='Men', color='blue')
plt.bar(x + 0.2, women_means, width=0.4, label='Women', color='pink')

plt.xticks(x, groups)#group labels
plt.xlabel("Groups")
plt.ylabel("Scores")
plt.title("Scores by Group and Gender")
plt.legend()
```

4 2,3,4,2,6

12. Write a Python program to create bar plots with error bars on the same figure. Sample Date
Mean velocity: 0.2474, 0.1235, 0.1737, 0.1824
Standard deviation of velocity: 0.3314, 0.2278, 0.2836, 0.2645

```python
mean_velocity = [0.2474, 0.1235, 0.1737, 0.1824]
std_dev = [0.3314, 0.2278, 0.2836, 0.2645]
labels = ['A', 'B', 'C', 'D']

plt.bar(labels, mean_velocity, yerr=std_dev, capsize=5, color='purple')
plt.xlabel("Categories")
plt.ylabel("Mean Velocity")
plt.title("Bar Chart with Error Bars")

```

13. Write a Python program to create bar plots with errorbars on the same figure. Attach a text
label above each bar displaying men means (integer value).
Sample Data
Mean velocity: 0.2474, 0.1235, 0.1737, 0.1824
Standard deviation of velocity: 0.3314, 0.2278, 0.2836, 0.2645

```python
labels = ['A', 'B', 'C', 'D']
mean_velocity = [0.25, 0.12, 0.17, 0.18]
std_dev = [0.33, 0.23, 0.28, 0.26]

plt.bar(labels, mean_velocity, yerr=std_dev, capsize=5, color='green')

plt.text(0, 0.27, "0.25", ha='center')
plt.text(1, 0.14, "0.12", ha='center')
plt.text(2, 0.19, "0.17", ha='center')
plt.text(3, 0.20, "0.18", ha='center')

plt.xlabel("Categories")
plt.ylabel("Mean Velocity")
plt.title("Bar Chart with Error Bars and Labels")
```
