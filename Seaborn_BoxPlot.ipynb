```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

1. Write a program to draw box plot of life expectancy by continent for a data set given in a
url
https://raw.githubusercontent.com/resbaz/r-novice-gapminder-files/master/data/gapmind
er-FiveYearData.csv

```python

url = "https://raw.githubusercontent.com/resbaz/r-novice-gapminder-files/master/data/gapminder-FiveYearData.csv"
df = pd.read_csv(url)
sns.boxplot(x="continent", y="lifeExp", data=df, palette={'Oceania' : "g", 'Africa' : "r", 'Europe' : 'b', 'Asia' : 'yellow', 'Americas' : 'skyblue'})
""
```
