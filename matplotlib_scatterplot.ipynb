```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
%matplotlib inline
```

1. Write a Python program to draw a scatter graph taking a random distribution in X and Y and
plotted against each other.

```python
x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x, y, edgecolors='r')

plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Scatter Plot with Random Distribution")
```

2. Write a Python program to draw a scatter plot with empty circles taking a random distribution
in X and Y and plotted against each other.

```python
x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x, y, facecolors='none', edgecolors='b')

plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Scatter Plot with Empty Circles")
```

3. Write a Python program to draw a scatter plot using random distributions to generate balls of
different sizes.

```python
x = np.random.rand(50)
y = np.random.rand(50)
sizes = np.random.rand(50) *100

plt.scatter(x, y, s=sizes, alpha=0.7, color='g')

plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Scatter Plot with Different Ball Sizes")
```

4. Write a Python program to draw a scatter plot comparing two subject marks of Mathematics
and Science. Use marks of 10 students.


```python
math_marks = [88, 92, 80, 89, 100, 80, 60, 100, 80, 34]
science_marks = [35, 79, 79, 48, 100, 88, 32, 45, 20, 30]

plt.scatter(math_marks, science_marks,facecolor = 'b' , color='r')


plt.xlabel("Math Marks")
plt.ylabel("Science Marks")
plt.title("Math vs Science Marks Scatter Plot")
```

5. Write a Python program to draw a scatter plot for three different groups camparing weights
and heights.

```python

group1_weights = np.random.randint(50, 80, 10)
group1_heights = np.random.randint(150, 180, 10)

group2_weights = np.random.randint(60, 90, 10)
group2_heights = np.random.randint(160, 190, 10)

group3_weights = np.random.randint(70, 100, 10)
group3_heights = np.random.randint(170, 200, 10)

plt.scatter(group1_weights, group1_heights, color='b', label="Group 1")
plt.scatter(group2_weights, group2_heights, color='g', label="Group 2")
plt.scatter(group3_weights, group3_heights, color='r', label="Group 3")

plt.xlabel("Weights (kg)")
plt.ylabel("Heights (cm)")
plt.title("Scatter Plot Comparing Weights and Heights")
plt.legend()

```
