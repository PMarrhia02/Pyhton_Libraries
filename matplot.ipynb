5. Write a Python program to plot two or more lines on same plot with suitable legends of each
line.

```python
x = [10, 20, 30]
y1 = [20, 40, 10]
y2 = [40, 10, 30]

plt.plot(x, y1, label="Line 1", marker='o', linestyle='-')
plt.plot(x, y2, label="Line 2", marker='s', linestyle='--')

plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.title("Two or More Lines on the Same Plot with Legends")
plt.legend()
plt.grid()
plt.show()
```

12. Write a Python program to create multiple types of charts.

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([10, 20, 15, 25, 30])

# Line Chart
plt.figure(figsize=(10, 6))

plt.subplot(2, 2, 1)  # rows, cols, index
plt.plot(x, y, 'r-o')
plt.title("Line Chart")

# Bar Chart
plt.subplot(2, 2, 2)
plt.bar(x, y, color='g')
plt.title("Bar Chart")

# Scatter Plot
plt.subplot(2, 2, 3)
plt.scatter(x, y, color='b')
plt.title("Scatter Plot")

# Pie Chart
plt.subplot(2, 2, 4)
explode = (0.4, 0, 0, 0, 0)
plt.pie(y, labels=x,explode=explode, autopct='%1.1f%%', colors=['r', 'g', 'b', 'y', 'c'], shadow=True)
plt.title("Pie Chart")

plt.tight_layout() #prevents overlapping

```

14. Write a Python program to display the grid and draw line charts of the closing value of
Alphabet Inc. between October 3, 2016 to October 7, 2016. Customized the grid lines with
rendering with a larger grid (major grid) and a smaller grid (minor grid).Turn on the grid but turn
off ticks.

```python


dates = ["03-10-16", "04-10-16", "05-10-16", "06-10-16", "07-10-16"]
closing_values = [772.56, 776.43, 776.47, 776.86, 775.08]

plt.plot(dates, closing_values, marker='o', linestyle='-', color='b', label="Closing Price")

plt.xlabel("Date")
plt.ylabel("Closing Value")
plt.title("Alphabet Inc. Closing Prices (Oct 3-7, 2016)")


plt.grid(True, which='major', linestyle='-', linewidth=0.8, color='black')                                                                                                      
plt.grid(True, which='minor', linestyle='--', linewidth=0.5, color='gray')                                                   


plt.tick_params(left=False, bottom=False) # remove ticks from  axis

plt.legend()
```

15. Write a Python program to create multiple plots.


```python
x = [1, 2, 3, 4, 5]
y1 = [1, 4, 9, 16, 25]  
y2 = [1, 8, 27, 64, 125] 

plt.figure(figsize=(20, 10))# creates fig with the spicific size width, height

# Subplot for(Square)
plt.subplot(1, 2, 1)  
plt.plot(x, y1, marker='o', color='b', label="y = x^2")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.title("Square Function")
plt.legend()
plt.grid(True)

#Subplot for (Cube)
plt.subplot(1, 2, 2)
plt.plot(x, y2, marker='s', color='r', label="y = x^3")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.title("Cube Function")
plt.legend()
plt.grid(True)

plt.tight_layout()  # Adjust layout to prevent overlap

```
